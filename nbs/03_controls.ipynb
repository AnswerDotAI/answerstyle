# Controls and feedback

> The skin improves affordance; the native element still does the work.

A real button submits a form, takes focus and reacts to a keyboard or pointer. A real input participates in form validation. A skin changes how those things read on the page without having to reproduce them. Applying `[btn]` to a link deliberately gives the link a button's visual weight without changing its navigation semantics. A link that merely leads somewhere can keep its normal link appearance.

## Actions

A raised primary button should look pushable before the pointer reaches it. Give a less important action the secondary face. Quiet and editable actions recede further: a small quiet control does not compete with the main button; the editable face marks a value that will become an input.

The same button can look pressed, expanded or busy according to its standard attributes. The skin does not supply the command, change `aria-pressed`, or turn a button into a link. Wrapper buttons default to `type="button"`; ordinary native buttons still follow their form defaults.

```
#| hide
#| asset styles.css
_css = r'''
@layer answerstyle.components {
@scope (:root, [ui]) to ([unstyle]) {
  :where(:scope, *)[btn] {
    --face: var(--color-accent);
    --ui-held-face: var(--v2);
    --ui-held-shadow: var(--bevel-in), var(--sunken);
    --ui-pad-block: var(--space-2);
    --ui-pad-inline: var(--space);
    --ui-press: 0px;
    position: relative;
    display: inline-flex; align-items: center; justify-content: center; gap: var(--space-2);
    padding: calc(var(--ui-pad-block) + var(--ui-press)) calc(var(--ui-pad-inline) - var(--ui-press))
      calc(var(--ui-pad-block) - var(--ui-press)) calc(var(--ui-pad-inline) + var(--ui-press));
    border: 1px solid transparent; border-radius: var(--radius);
    background: var(--face); color: var(--color-on-accent); box-shadow: var(--bevel), var(--raised);
    font: inherit; font-weight: var(--weight1); line-height: var(--leading-1);
    text-align: center; text-decoration: none; cursor: pointer;
    transition: background var(--motion) var(--ease), box-shadow var(--motion) var(--ease), padding calc(var(--motion) / 2) var(--ease), opacity var(--motion) var(--ease);

    & > :where(span) { display: inline-flex; align-items: center; justify-content: inherit; gap: var(--space-2); }
    &:where(:hover:not(:disabled)) { --face: var(--v1); }
    &:where(:active:not(:disabled)) { --ui-press: calc(1px * var(--relief)); }
    &:where(:focus-visible) { outline: var(--ring); outline-offset: 2px; }
    &:where([btn-size=sm]) { --ui-pad-block: var(--space-3); --ui-pad-inline: var(--space-2); font-size: var(--text-1); }
    &:where([btn-size=lg]) { --ui-pad-block: var(--space-1); --ui-pad-inline: var(--space1); font-size: var(--text1); }
    &:where([btn-variant=secondary]) {
      --face: var(--color-surface); --ui-held-face: var(--color-surface); color: var(--color-text); border-color: var(--color-border);
      &:where(:hover:not(:disabled)) { --face: var(--v-3); }
      &:where([aria-pressed=true], [aria-expanded=true]) { --ui-held-face: var(--v-3); color: var(--v2); }
    }
    &:where([btn-variant=quiet], [btn-variant=editable]) {
      --face: transparent; --ui-held-face: color-mix(in oklch, currentColor 12%, transparent); --ui-held-shadow: none;
      --ui-press: 0px; color: inherit; box-shadow: none;
      &:where(:hover:not(:disabled)) { --face: var(--tint); }
    }
    &:where([btn-variant=editable]) {
      --ui-pad-inline: var(--space-1); inline-size: 100%; justify-content: start; border-color: var(--n); border-style: dashed;
      font-weight: inherit; text-align: start; cursor: text;
    }
    &:where(:is(:active, [aria-busy=true], [aria-pressed=true], [aria-expanded=true]):not(:disabled)) {
      --face: var(--ui-held-face); box-shadow: var(--ui-held-shadow);
    }
    &:where(:disabled) { opacity: .5; box-shadow: none; cursor: default; }
    &:where([aria-busy=true]) {
      -webkit-text-fill-color: transparent;
      & > :where(span, svg) { opacity: 0; }
      &::after {
        content: ''; position: absolute; inset: 0; margin: auto; inline-size: 1.25em; block-size: 1.25em;
        border: 2px solid currentColor; border-inline-end-color: transparent; border-radius: var(--radius-full);
        animation: ui-spin calc(var(--motion) * 5) linear infinite;
      }
    }
    &:where([btn-icon]) { --ui-pad-inline: var(--ui-pad-block); }
    &:where([btn-icon]) > :where(span) { inline-size: 1.25em; block-size: 1.25em; }
    @media (prefers-reduced-motion: reduce) {
      transition: none;
      &:where([aria-busy=true])::after { animation-duration: 2s; }
    }
  }
}
}
'''
```

## Wells and groups

An input reads as a place to type: it has a quieter face and the opposite relief from a button. Send the label and description relationship through ordinary HTML. The label's `for` points to the control's `id`; `aria-describedby` points to the nearby hint. Validation can use the browser's state or a server-set `aria-invalid`.

A cluster spaces separate controls. A group visually joins neighbouring controls and preserves one overall outline for a focused input. The inset variant places a small action inside the well; it isn't a general row layout.

## Surfaces, status and data

A card is a surface, not a required heading/body/footer skeleton. Add a stack when its contents actually need one. A badge tells the reader something but does not look like a command. A spinner can explain that work is underway, and a styled progress element still carries normal progress semantics. Use an alert for an enduring notice, not a brief state of a button.

`table[table]` is an application data table with native headings, a bottom caption, and row rules. Use `num` on comparable figures to align them at the end with tabular digits. A labelled, focusable `[table-frame]` adds a nonwrapping ledger treatment with a scrolling boundary. Long-form prose has a different table treatment and keeps ordinary native table markup.

```
#| hide
#| asset styles.css
_css = r'''
@layer answerstyle.components {
@scope (:root, [ui]) to ([unstyle]) {
  :where(:scope, *)[card] { --face: var(--color-surface); padding: var(--space); background: var(--face); border: 1px solid var(--color-border); border-radius: var(--radius); box-shadow: var(--raised); }
  :where(:scope, *)[badge] { display: inline-block; padding: .125em .5em; border-radius: var(--radius-1); background: var(--v-3); color: var(--v2); font-size: var(--text-1); font-weight: var(--weight1); line-height: var(--leading-1); }
  :where(:scope, *)[badge][badge-variant=secondary] { background: var(--color-surface); color: var(--color-text); box-shadow: inset 0 0 0 1px var(--color-border); }
  :where(:scope, *)[avatar] { display: inline-grid; place-items: center; vertical-align: middle; flex: none;
    inline-size: var(--avatar-size, 2.5rem); block-size: var(--avatar-size, 2.5rem); border-radius: var(--radius-full);
    object-fit: cover; background: var(--v-3); color: var(--v2); font-size: calc(var(--avatar-size, 2.5rem) * .4);
    font-weight: var(--weight1); line-height: 1; }

  [alert] { display: flex; align-items: flex-start; gap: var(--space); padding: var(--space); border: 1px solid var(--v-2);
    border-radius: var(--radius); background: var(--v-3); color: var(--v2); }
  [alert] > [stack] { --gap: var(--space-2); flex: 1; min-inline-size: 0; }
  [progress] { inline-size: 100%; block-size: var(--space-2); accent-color: var(--color-accent); }
  [spinner] { display: inline-flex; align-items: center; gap: var(--space-2); }
  [spinner]::before { content: ""; inline-size: 1em; block-size: 1em; flex: none; border: 2px solid currentColor;
    border-inline-end-color: transparent; border-radius: var(--radius-full); animation: ui-spin calc(var(--motion) * 5) linear infinite; }
  [skeleton] { display: block; inline-size: 100%; block-size: 1em; background: var(--n-1); border-radius: var(--radius-1); }
  @media (prefers-reduced-motion: reduce) { [spinner]::before { animation: none; } }

  :where(:scope, *)[table-frame] {
    overflow-x: auto; contain: inline-size; background: var(--color-background); border: 1px solid var(--color-border);
    border-radius: var(--radius); box-shadow: var(--raised);
  }
  :where(:scope, *)[table-frame]:focus-visible { outline: var(--ring); outline-offset: 2px; }
  table[table]:where(:scope, *) { inline-size: 100%; border-collapse: separate; border-spacing: 0; }
  table[table] :where(th, td) { padding: var(--space-2) var(--space); text-align: start; }
  table[table] :where(thead th) { color: var(--color-text); font-weight: var(--weight1); border-block-end: 1px solid var(--color-border); }
  table[table] :where(tbody tr + tr > *, tfoot tr > *) { border-block-start: 1px solid var(--color-border); }
  table[table] :where(tfoot th) { font-weight: var(--weight1); }
  table[table] :where(caption) { caption-side: bottom; padding: var(--space-2) var(--space); color: var(--color-muted); font-size: var(--text-1); text-align: start; }
  table[table] :is(th, td)[num] { text-align: end; font-variant-numeric: tabular-nums; }
  [table-frame] table[table] :where(th, td) { white-space: nowrap; }
  [table-frame] table[table] :where(thead th) { background: var(--color-surface); }
}
}
'''
```
