# AnswerStyle

> Shared styles and browser behavior for semantic HTML

AnswerStyle's HTML contract is the same in a Svelte component, a FastHTML page, and a hand-written web page. Mark a native element to give it a skin. Use CSS custom properties to set the palette, type, spacing, and relief where you need a different appearance.

The notebooks derive the colors and type and show how layout, surfaces, and controls compose. Run the preview cells to try the examples at their intended widths. Open the browser assets from either the npm or Python package.
