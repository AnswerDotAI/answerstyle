## Collapse when the actual row no longer fits

The same navbar can fit a translated label in one region and run out of room in another. Measure its real row instead of guessing from a fixed breakpoint. The wide and narrow states use the same labelled navigation region and links. The toggle controls that region, not an ARIA menu. Without the controller the links still appear and wrap.

The compact layout puts the links in a second row and keeps brand, toggle and end actions available. A command toolbar and breadcrumb use the same measurement, but own their own compact content and focus behavior.

## Modal, popover and menu are different

A modal interrupts work and owns focus until the operation is completed or cancelled. Use a real `dialog` and `showModal()`. A dropdown is a native non-modal popover containing ordinary links, buttons and form controls. Tab moves between its controls; Escape and clicking outside dismiss it. It is not an ARIA application menu. Nested choices use a native `details` group instead of a cascading flyout.

CSS anchor positioning places a popover beside its trigger and flips it to use the available space. `mountPopover(trigger, panel)` supplies the same placement on engines without anchors and keeps `aria-expanded` in sync. It returns a cleanup function. Native popovers use the browser's top layer without moving the DOM to an unrelated theme.

A tooltip describes an already identified control through `aria-describedby`. CSS shows the description on hover or keyboard focus, including while crossing into the description. Unlike a popover it does not own focus or dismiss with Escape. No tooltip controller is needed.

## Disclosures, tabs and transient feedback

A disclosure is native `details` with a summary. Name multiple details elements alike when only one should stay open.

Tabs need explicit selection and panel relationships. `mountTabs(root, {activationMode, onValueChange})` handles arrow keys, Home/End and disabled tabs. Automatic activation selects a tab when its arrow key receives focus. Manual activation waits for Enter, Space or a click. The controller's `setValue(value)` updates selection; `dispose()` removes its listeners. Use vertical orientation when the tab list runs down the side.

`mountToolbar(root)` gives a command row one Tab stop and Left/Right traversal. It leaves dropdown contents in normal Tab order. Application commands and responsive presentation remain outside that helper. `mountHoverPopover(trigger, panel)` adds mouse opening and a short crossing delay, plus ArrowDown to open and focus settings. A touch tap also opens the panel. The trigger's click action stays independent, such as a reasoning toggle that also opens its model settings on touch.

A transient toast is not an alert. Its region remains present so new messages can be announced. The skin supplies a lifetime that pauses on hover or focus; the host removes the toast from its state or DOM when the lifetime ends. Urgent feedback during a modal task belongs inside that modal.
