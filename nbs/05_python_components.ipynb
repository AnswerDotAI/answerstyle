# Python components

> Two FastHTML components built on AnswerStyle's HTML attributes.

These are candidate AnswerUI components. Their styles come from AnswerStyle; the Python code only builds HTML.

```
from fasthtml.common import Button, Section, H2, P, Div, to_xml

from answerstyle.preview import preview
```

A shared constructor sets the default attributes and translates Python argument names when needed. Children and all other attributes go straight to the native tag.

```
def component(tag, /, *, rename=None, **defaults):
    names = rename or {}
    def render(*c, **kwargs): return tag(*c, **(defaults | {names.get(k, k): v for k,v in kwargs.items()}))
    return render
```

## Button

Use `Btn` for a styled control. It defaults to `type="button"` to avoid submitting a surrounding form; use `type='submit'` for its submit action. FastHTML's `Button` stays unstyled by AnswerStyle's button skin.

```
Btn = component(Button, btn=True, type='button', rename={'variant': 'btn_variant'})
```

```
edit = Btn('Edit', variant='secondary')
print(to_xml(edit, Button('Native')))
```

## Card

The card skin and the layout are separate attributes. `Card` uses a `section` and passes `stack=True` to the same element when you want a vertical layout. For a different semantic element, use the skin directly, for example `Article(..., card=True)`.

```
Card = component(Section, card=True)
```

```
draft = Card(
    H2('Offer letter', id='offer-letter'),
    P('Draft for review.'),
    Div(edit, Btn('Export'), cluster=True),
    stack=True, aria_labelledby='offer-letter')

preview(to_xml(draft), height='12rem')
```

A Svelte template can use the same `<section card stack>` and put its actions in a `<div cluster>`. The edit action is `<button btn type="button" btn-variant="secondary">Edit</button>`.

## Native Dropdown and Tooltip

The [FastHTML example](../examples/native.py) registers two Python constructors without registering component CSS or JavaScript. `Dropdown` emits an ordinary popover of links and buttons. AnswerStyle positions it, and the browser handles focus and dismissal. `Tooltip` connects its description to the trigger with `aria-describedby` and uses the AnswerStyle CSS.

With FastHTML and AnswerUI installed, run `uvicorn examples.native:app --port 5024` from the AnswerStyle checkout.
