# Layout without outer margins

> Arrange the relationships between children, not the children themselves.

The same control goes in a narrow form, a toolbar or a wide panel. It can own its internal padding in all three. It cannot tell how far away its neighbours belong. Put the gap on the parent, and the last item never leaves a stray margin at the edge.

The layout is an attribute on the HTML you already need: a `section[stack]`, a `nav[cluster]`, or a `ul[grid]`. A Svelte component or a Python constructor can emit that markup. Neither needs to insert a second element or generate a rule.

```
from answerstyle.preview import preview
```

## Stack, Cluster and Grid

Stack puts items above one another and stretches them to the available width. Cluster lets them keep their natural widths, centres them on the cross axis and wraps when needed. Grid gives them equal-width columns: as many columns as fit at the requested minimum, with a narrow single column allowed to shrink. Each layout has its own `--gap`, so spacing on a surrounding grid never changes the spacing inside a stack.

Sometimes one child should fill the space others leave. `[grow]` gives that child the remaining width in a cluster or remaining height in a stack. In a row, a long name can wrap instead of squeezing out the actions.

```
#| hide
#| asset styles.css
_css = r'''
@layer answerstyle.layout {
  @scope (:root, [ui]) to ([unstyle]) {
    :where(:scope, *):is([stack], [cluster], [grid]) { --gap: var(--space); }
    :where(:scope, *)[stack] { display: flex; flex-direction: column; gap: var(--gap); }
    :where(:scope, *)[cluster] {
      --cluster-wrap: wrap; --cluster-align: center; --cluster-justify: start;
      display: flex; flex-wrap: var(--cluster-wrap); align-items: var(--cluster-align); justify-content: var(--cluster-justify); gap: var(--gap);
    }
    :where(:scope, *)[grid] {
      --grid-min: 16rem; display: grid; gap: var(--gap);
      grid-template-columns: repeat(auto-fit, minmax(min(100%, var(--grid-min)), 1fr));
    }
    :where(:scope, *)[grid] > * { min-inline-size: 0; }
    :where(:scope, *):is([stack], [cluster]) > [grow] { flex: 1; min-inline-size: 0; min-block-size: 0; }
    :where(:scope, *):is(ul, ol):is([stack], [cluster], [grid]) { padding: 0; list-style: none; }
    :where(:scope, *)[toolbar] { display: flex; align-items: center; gap: var(--space-2); min-inline-size: 0; }
  }
}
'''
```

```
member = '''<li cluster>
  <div stack grow style="--gap:var(--space-3)"><strong>Ada Lovelace</strong><small>Workspace owner</small></div>
  <button type="button" btn btn-variant="secondary">Edit</button>
</li>'''
preview(f'''<section card style="max-inline-size:26rem">
  <ul stack>{member}{member.replace('Ada Lovelace','Grace Hopper').replace('Workspace owner','Editor')}</ul>
</section>''', height='12rem')
```

## Flow follows the text

A brief help section reads as paragraphs with closely related headings. Use Flow when you need that rhythm but not an article's prose styles. The parent owns the row gap; a noninitial heading gets extra separation above it. A nested flow may tighten its own rows without changing the gap its parent assigns to it. For full articles, `[prose]` also sets list, code and table typography.

```
#| hide
#| asset styles.css
_css = r'''
@layer answerstyle.layout {
  @scope (:root, [ui]) to ([unstyle]) {
    :where(:scope, *)[flow] { --flow-space: .5em; display: grid; align-content: start; row-gap: var(--flow-space); }
    :where(:scope, *)[flow] > :is(h1, h2, h3, h4, h5, h6):not(:first-child) { margin-block-start: .5em; }
  }
}
'''
```

```
preview('''
<article card flow>
  <h2>Sharing a workspace</h2>
  <p>Invite your colleagues and decide what they can edit.</p>
  <section flow style="--flow-space:.2em">
    <h3>Permissions</h3>
    <p>Owners can invite others.</p>
    <p>Editors can change documents.</p>
  </section>
  <p>Changes apply immediately.</p>
</article>
''', height='16rem')
```

## Small layout utilities

Use a semantic arrangement when it describes the job. For a one-off layout, a few utilities combine directly: `flex items-center gap-2`. Numbered padding and gap utilities use the existing spacing steps. Step 1 is one quarter of the base; steps 2, 3 and 4 are one half, three quarters and the whole base. `px` and `py` use logical inline and block axes.

`[responsive-layout]` is an explicit opt-in for a wide and compact presentation in the same container. It hides alternatives; it does not invent a new display value for either of them.

```
#| hide
#| asset styles.css
_css = r'''
@layer answerstyle.utilities {
@scope (:root, [ui]) to ([unstyle]) {
  .ui-visually-hidden { position: absolute; inline-size: 1px; block-size: 1px; padding: 0; overflow: hidden; clip-path: inset(50%); white-space: nowrap; }
  .ui-hidden, :where(:scope, *)[hidden]:not([hidden=until-found]) { display: none !important; }

  .p-1 { padding: var(--space-3); }
  .p-2 { padding: var(--space-2); }
  .p-3 { padding: var(--space-1); }
  .p-4 { padding: var(--space); }
  .px-1 { padding-inline: var(--space-3); }
  .px-2 { padding-inline: var(--space-2); }
  .px-3 { padding-inline: var(--space-1); }
  .px-4 { padding-inline: var(--space); }
  .py-1 { padding-block: var(--space-3); }
  .py-2 { padding-block: var(--space-2); }
  .py-3 { padding-block: var(--space-1); }
  .py-4 { padding-block: var(--space); }
  .gap-1 { gap: var(--space-3); }
  .gap-2 { gap: var(--space-2); }
  .gap-3 { gap: var(--space-1); }
  .gap-4 { gap: var(--space); }
  .flex { display: flex; }
  .flex-col { flex-direction: column; }
  .flex-wrap { flex-wrap: wrap; }
  .items-center { align-items: center; }
  .justify-between { justify-content: space-between; }
  .justify-end { justify-content: end; }
  .flex-1 { flex: 1; }
  .flex-none { flex: none; }
  .min-w-0 { min-width: 0; }
  [responsive-layout] { container: ui-layout / inline-size; }
  @container ui-layout (width < 48rem) { .ui-wide-only { display: none !important; } }
  @container ui-layout (width >= 48rem) { .ui-compact-only { display: none !important; } }
}
}
'''
```
